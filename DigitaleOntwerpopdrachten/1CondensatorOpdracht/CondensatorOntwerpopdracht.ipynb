##### Vraag 1a Het potentiaal rond een puntlading

1. Bepaal welke index correspondeert met $z = 0$ m.
2. Plot de log van het potentiaal tegen de log van de straal.
3. Fit de log van het potentiaal met de log van de straal.
4. Print het exponent op een nette manier.

````
from scipy.optimize import curve_fit
````

````
#index bepalen voor z= 0 m
idx = np.argmin(np.abs(z-0)) #kijken bij welke indexhet verschil tussen z en 0 minimaal is
print(f"index: {idx}, z: {z[idx]:.3f} m")

#natuurlijk logaritme nemen van arrays
log_V = np.log(V[idx,:])
log_V = log_V[1:] #first value sliced away because it is inf
log_r = np.log(r)[1:] #first value sliced away because it is inf


#fitfunctie en fit maken
def fitfunc(x,a,b):
    return a*x + b

popt, pcov = curve_fit(fitfunc,log_V, log_r)

xfit = np.linspace(np.min(log_r), np.max(log_r),1000)
yfit = fitfunc(xfit, *popt)

#fitfunctie plotten
plt.figure()
plt.plot(log_r,log_V,".",label=r"$\ln(V)$ tegen $\ln(r)$ voor z=0 m") #potentiaal voor z = 0 plotten
plt.plot(xfit,yfit,"r--",label = "fitfunctie")
plt.xlabel(r"$r$ (m)")
plt.ylabel(r"$V$ (V)")
plt.legend()
plt.show()

#exponent printen
print(f"exponent: {popt[0]:.2f} \u00B1 {np.sqrt(pcov[0,0]):.2f}")
````

#### Vraag 2h
Het elektrische veld een afstand $z$ boven het midden van een uniform geladen schijf met een straal $R$ en oppervlakte lading $\sigma$ wordt gegeven door:
$$ \mathbf{E} = \frac{\sigma}{2\epsilon_0} \bigg[ 1- \frac{z}{\sqrt{R^2 + z^2}}\bigg] \hat{\mathbf{z}}$$
Zie probleem 2.6 in het boek van Griffith.

1. Wat moet de hoogte $z$ zijn boven de schijf zodat het elektrische veld door de schijf 10% is in vergelijking met dat van een oneindig geladen plaat?

Voeg hieronder je analytische uitwerking toe.

##### Antwoord 2h
Het elektrische veld boven het midden van een uniform geladen schijf is

$$
E(z)=\frac{\sigma}{2\varepsilon_0}\left(1-\frac{z}{\sqrt{R^2+z^2}}\right)
$$

en voor een oneindige plaat geldt

$$
E_\infty=\frac{\sigma}{2\varepsilon_0}.
$$

Hoogte $z$ waarvoor het veld van de schijf nog maar $10\%$ is van dat van de oneindige plaat

$$
E(z)=0.1E_\infty.
$$

Invullen geeft

$$
1-\frac{z}{\sqrt{R^2+z^2}}=0.1
\quad \Rightarrow \quad
\frac{z}{\sqrt{R^2+z^2}}=0.9.
$$

Kwadrateren levert

$$
\frac{z^2}{R^2+z^2}=0.81.
$$

Oplossen geeft

$$
z \approx 2.06R
$$

1. Vind de formule voor de capaciteit van een vlakke plaat condensator en een cilinder condensator, onder de aanname dat de afstand tussen de platen veel kleiner is dan het oppervlakte van de condensator (vlakke plaat aanname).
2. Schrijf een python functie die de relevante parameters als input neemt en de capaciteit van een vlakke plaat condensator als output heeft.
3. Vind de formule voor de capaciteit van een cilinder condensator onder de aanname dat de afstand tussen de binnen en buiten wand klein is.
4. Schrijf een python functie die de relevante parameters als input neemt en de capaciteit van een cilinder condensator als output heeft.

````
import numpy as np
import matplotlib.pyplot as plt

epsilon0 = 8.854e-12  # permittiviteit van vacuum (F/m)

def vlakkeplaat(A, d, epsilon_r):
    C = (epsilon0 * epsilon_r * A) / d
    return C

def cilinder(r_in, r_uit, L, epsilon_r):
    C = (2 * np.pi * epsilon0 * epsilon_r * L) / np.log(r_uit / r_in)
    return C
````

Formule voor vlakke plaat condensator

$$
C=\frac{\varepsilon_0 \, \varepsilon_r \, A}{d}
$$

Met $C$ de capaciteit, $\varepsilon_0$ de permitiviteit van vacuüm, $\varepsilon_r$ de relatieve permitiviteit van het materiaal tussen de platen, $A$ de oppervlakte van één plaat en $d$ de afstand tussen de platen (in meter).

Formule voor cilindrische condensator

$$
C=\frac{2\pi \,\varepsilon_0 \,\varepsilon_r \,L}{\ln\!\left(\frac{r_{\text{uit}}}{r_{\text{in}}}\right)}
$$

Met $r_{\text{uit}}$ de buitenstraal van de cilinder, $r_{\text{in}}$ de binnenstraal en $L$ de lengte van de cilinder.
